## 2. Quick introduction to PycWB

> PycWB is a project to Pythonize cWB, but is not just rewriting the C++/ROOT code to Python. PycWB restructures the pipeline into modern framework.

 - PycWB introduced module framework
 - The core code still relies on cWB C++ code through pyROOT
 - The main workflow is completed, more functionalities are under activate developement.
